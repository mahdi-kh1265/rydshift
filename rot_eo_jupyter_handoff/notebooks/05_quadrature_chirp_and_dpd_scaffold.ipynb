# 05 Quadrature chirp and DPD scaffold

This notebook connects RF transfer functions to actual Ex/Ey quadrature errors and eventually optical spurs.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT / "src"))

import numpy as np
import matplotlib.pyplot as plt
from rot_eo_model.dpd import safe_inverse_transfer, predistort_spectrum, quadrature_error_metrics
```

```python
# Synthetic transfer functions for x/y channels. Replace with measured S-params or fitted network response.
f = np.linspace(6e6, 102e6, 2000)
Hx = 1/(1+1j*f/80e6)
Hy = 0.9*np.exp(1j*np.deg2rad(7))/(1+1j*f/70e6)
metrics = quadrature_error_metrics(Hx, Hy)
plt.figure()
plt.plot(f/1e6, metrics["amp_ratio_y_over_x"])
plt.xlabel("Frequency [MHz]"); plt.ylabel("|Hy|/|Hx|"); plt.grid(True)
plt.figure()
plt.plot(f/1e6, np.rad2deg(metrics["phase_error_rad"]))
plt.xlabel("Frequency [MHz]"); plt.ylabel("Phase error from 90° [deg]"); plt.grid(True)
```

```python
# Example linear inverse predistortion gain requirements
invx = safe_inverse_transfer(Hx, floor=0.05)
invy = safe_inverse_transfer(Hy, floor=0.05)
plt.figure()
plt.plot(f/1e6, 20*np.log10(np.abs(invx)), label="x inverse gain")
plt.plot(f/1e6, 20*np.log10(np.abs(invy)), label="y inverse gain")
plt.xlabel("Frequency [MHz]"); plt.ylabel("Predistortion gain [dB]"); plt.legend(); plt.grid(True)
```
